### Gaussian mixture modeling of Iris dataset

The code below loads the Iris dataset that we have looked at previously. Perform mixture modeling under the assumption of Gaussian distribution. Experiment with different number of components, initialization methods, and number of iterations.

```
from ucimlrepo import fetch_ucirepo
```

```
# Iris starter: 2 of the 4 features (sepal length, sepal width),
# all 3 classes. Students fit a Gaussian mixture model and experiment
# with the number of components, initialization, and iterations.
iris = fetch_ucirepo(id=53)

X_iris = iris.data.features.iloc[:, :2]
X_1_name = "sepal length (cm)"
X_2_name = "sepal width (cm)"
y_iris = np.zeros(150)
y_iris[50:100] = 1
y_iris[100:150] = 2
y_names = np.unique(iris.data.targets)

fig_iris_gmm, ax_iris_gmm = plt.subplots(figsize=(6, 6))
for class_i, class_name in enumerate(y_names):
    ax_iris_gmm.scatter(
        X_iris.iloc[np.where(y_iris == class_i)[0], 0],
        X_iris.iloc[np.where(y_iris == class_i)[0], 1],
        label=class_name,
    )
ax_iris_gmm.set_xlabel(X_1_name)
ax_iris_gmm.set_ylabel(X_2_name)
ax_iris_gmm.legend()

display(fig_iris_gmm)
plt.close(fig_iris_gmm)
```
