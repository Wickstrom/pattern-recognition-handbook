### Where momentum shines: an ill-conditioned cost surface

The data below is two overlapping Gaussians whose covariance is strongly
*anisotropic* (a narrow direction and a wide direction, rotated off the
axes). This makes the cost surface an elongated valley: plain gradient
descent zig-zags across the narrow direction and advances slowly along the
wide one. This is exactly the setting where momentum helps by averaging out
the oscillations. Train your MLP with and without momentum and compare the
training curves and the number of iterations to converge.


```
# Rotated anisotropic Gaussians: one covariance eigen-direction is much
# longer than the other, so the cost surface is a steep, elongated
# valley. Plain GD bounces across the narrow direction; momentum damps it.
n_elong = 80
rng_elong = np.random.default_rng(0)
theta_rot = np.pi / 4
R_rot = np.array([[np.cos(theta_rot), -np.sin(theta_rot)],
                  [np.sin(theta_rot), np.cos(theta_rot)]])
S_elong = np.diag([0.15, 3.0])
cov_elong = R_rot @ S_elong @ R_rot.T
mu1_elong = np.array([0.0, 0.0])
mu2_elong = np.array([1.8, 1.8])

X1_elong = rng_elong.multivariate_normal(mu1_elong, cov_elong, n_elong)
X2_elong = rng_elong.multivariate_normal(mu2_elong, cov_elong, n_elong)
X_elong = np.vstack([X1_elong, X2_elong])
y_elong = np.concatenate([np.zeros(n_elong), np.ones(n_elong)])

fig_elong, ax_elong = plt.subplots(figsize=(6, 6))
ax_elong.scatter(X_elong[:, 0], X_elong[:, 1], c=y_elong, cmap="viridis",
                 edgecolor="k", s=50)

display(fig_elong)
plt.close(fig_elong)
```
