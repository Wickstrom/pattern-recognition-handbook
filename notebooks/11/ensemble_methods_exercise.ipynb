```
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split as _tts

X01, y01 = make_moons(n_samples=300, noise=0.45, random_state=3)
y_pm1 = np.where(y01 == 0, -1, 1)  # AdaBoost expects labels in {-1, +1}
X_train, X_test, y_train, y_test = _tts(X01, y_pm1, test_size=0.5, random_state=3)

plt.figure(1)
plt.scatter(X_train[:, 0], X_train[:, 1], c=y_train, cmap="coolwarm", edgecolor="k", s=30)
plt.title("Training set: noisier Two Moons (noise=0.45)")
plt.show()

# Your AdaBoost implementation and train/test error plot go here.
```
