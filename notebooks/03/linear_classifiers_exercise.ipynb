### Problem 2.12 from the book

Repeat problem 2.12 from the book, but using the MSE, SE2, and Perceptron classifier. How do they compare to the density-based classifiers from the previous sessions?

For the Perceptron algorithm to converge, the data needs to be linearly separable. The easiest way to accomplish this is to change the mean of one of the classes until the two classes are completely separable.

```
# Problem 2.12 starter: two overlapping Gaussians with the same
# covariance. The Bayes-optimal boundary is the perpendicular
# bisector of the line between the means, but MSE/LS won't recover
# that exactly — that's the point of the exercise.
number_of_samples_p212 = 100

mu1_p212 = np.array([1, 1])
mu2_p212 = np.array([1.5, 1.5])
sigma_p212 = np.array([[0.2, 0.0], [0.0, 0.2]])

x_train_1_p212 = np.random.multivariate_normal(mu1_p212, sigma_p212, number_of_samples_p212)
x_train_2_p212 = np.random.multivariate_normal(mu2_p212, sigma_p212, number_of_samples_p212)

fig_p212, ax_p212 = plt.subplots(figsize=(8, 8))
ax_p212.scatter(
    x_train_1_p212[:, 0], x_train_1_p212[:, 1],
    s=120, facecolors="none", edgecolors="black", linewidth=3.0, label="Class 1",
)
ax_p212.scatter(
    x_train_2_p212[:, 0], x_train_2_p212[:, 1],
    s=120, facecolors="none", edgecolors="blue", linewidth=3.0, label="Class 2",
)
ax_p212.set_xlabel("x1")
ax_p212.set_ylabel("x2")
ax_p212.legend()

display(fig_p212)
plt.close(fig_p212)
```

### Wine classification with LS classifiers

This problem is focused on multi-class classification of wine types based on 13 features. Below, we have simplified the problem by extracting 2 features. Take this as a starting point, but when you feel confident in your implementation, train your classifier on all 13 features. Does performance change?

```
from sklearn.datasets import load_wine
```

```
# Wine starter: 2 of the 13 features (alcohol, malic acid), all 3
# classes. Targets are 0/1/2; legend is shifted to 1/2/3 to match
# the original notebook's "Class N" labeling.
wine_data = load_wine()

X_wine = wine_data.data[:, :2]
feature_1_name_wine = "alcohol"
feature_2_name_wine = "malic acid"
y_wine = wine_data.target
y_names_wine = np.unique(y_wine)
colors_wine = ["black", "blue", "red"]

fig_wine, ax_wine = plt.subplots(figsize=(5, 5))
for class_value, color in zip(y_names_wine, colors_wine):
    ax_wine.scatter(
        X_wine[y_wine == class_value, 0],
        X_wine[y_wine == class_value, 1],
        s=120, facecolors="none",
        edgecolors=color, linewidth=3.0,
        label=f"Class {class_value+1}",
    )
ax_wine.set_xlabel(feature_1_name_wine)
ax_wine.set_ylabel(feature_2_name_wine)
ax_wine.legend()

display(fig_wine)
plt.close(fig_wine)
```

### Beyond a single line: non-linearly separable data

Every classifier above (MSE, SE2, Perceptron) draws one straight decision
boundary, so none of them can separate classes that *wrap around* each other.
Fit your classifiers on the two datasets below and confirm that they cannot
perfectly separate either one, no matter how you shift/rescale the features.
This is exactly the gap the kernel (lecture 6) and neural-network (lecture 4)
methods are designed to close.


```
from sklearn.datasets import make_circles

# Two circles: class 1 forms a ring that fully surrounds class 2.
# No single straight line can separate them.
X_circles, y_circles = make_circles(n_samples=400, factor=0.5, noise=0.08, random_state=0)

fig_circles, ax_circles = plt.subplots(figsize=(6, 6))
ax_circles.scatter(
    X_circles[y_circles == 0, 0], X_circles[y_circles == 0, 1],
    s=40, facecolors="none", edgecolors="black", linewidth=1.5, label="Class 1",
)
ax_circles.scatter(
    X_circles[y_circles == 1, 0], X_circles[y_circles == 1, 1],
    s=40, facecolors="none", edgecolors="blue", linewidth=1.5, label="Class 2",
)
ax_circles.set_xlabel("x1")
ax_circles.set_ylabel("x2")
ax_circles.legend()

display(fig_circles)
plt.close(fig_circles)
```

### XOR: classes that alternate corners

This is the classic XOR layout: class 1 occupies the bottom-left and top-right
corners, class 2 the other two. The classes are completely non-overlapping, yet
no single line can separate them — the pattern is only separable by a
non-linear (e.g. neural-network) boundary.


```
# Four Gaussian blobs in the XOR arrangement on the unit square.
# Class 0: corners (0,0) and (1,1); Class 1: corners (0,1) and (1,0).
n_per_blob_xor = 100
sigma_xor = np.array([[0.05, 0.0], [0.0, 0.05]])
corner_sets_xor = [
    (np.array([0.0, 0.0]), 0),
    (np.array([1.0, 1.0]), 0),
    (np.array([0.0, 1.0]), 1),
    (np.array([1.0, 0.0]), 1),
]

X_xor = np.vstack([
    np.random.multivariate_normal(mu, sigma_xor, n_per_blob_xor) for mu, _ in corner_sets_xor
])
y_xor = np.concatenate([
    np.full(n_per_blob_xor, label) for _, label in corner_sets_xor
])

fig_xor, ax_xor = plt.subplots(figsize=(6, 6))
ax_xor.scatter(X_xor[y_xor == 0, 0], X_xor[y_xor == 0, 1],
               s=40, facecolors="none", edgecolors="black", linewidth=1.5, label="Class 1")
ax_xor.scatter(X_xor[y_xor == 1, 0], X_xor[y_xor == 1, 1],
               s=40, facecolors="none", edgecolors="blue", linewidth=1.5, label="Class 2")
ax_xor.set_xlabel("x1")
ax_xor.set_ylabel("x2")
ax_xor.legend()

display(fig_xor)
plt.close(fig_xor)
```

### Sensitivity to outliers

Here the two classes are cleanly separated, but each class carries a few
*outliers* far from its mean. The Perceptron still converges (the data is
linearly separable), while MSE/LS can be dragged badly off the optimal line by
the squared errors of those outlier points. Compare the decision boundary with
and without the outliers removed.


```
# Two well-separated Gaussians, each with a handful of far-out points
# appended. Same covariance, so the Bayes-optimal line is the midpoint
# perpendicular bisector — but the outliers will pull MSE/LS away from it.
n_core_out = 80
n_outliers_out = 4
mu_core_out = np.array([-1.2, 0.0])
mu2_core_out = np.array([1.2, 0.0])
sigma_core_out = np.array([[0.2, 0.0], [0.0, 0.2]])

x1_core_out = np.random.multivariate_normal(mu_core_out, sigma_core_out, n_core_out)
x2_core_out = np.random.multivariate_normal(mu2_core_out, sigma_core_out, n_core_out)
x1_out_out = mu_core_out + 4.5 * np.ones((n_outliers_out, 2)) \
    + np.random.multivariate_normal([0, 0], 0.05 * np.eye(2), n_outliers_out)
x2_out_out = mu2_core_out - 4.5 * np.ones((n_outliers_out, 2)) \
    + np.random.multivariate_normal([0, 0], 0.05 * np.eye(2), n_outliers_out)

X_out = np.vstack([x1_core_out, x1_out_out, x2_core_out, x2_out_out])
y_out = np.concatenate([
    np.zeros(n_core_out + n_outliers_out), np.ones(n_core_out + n_outliers_out),
])

fig_out, ax_out = plt.subplots(figsize=(7, 6))
ax_out.scatter(X_out[y_out == 0, 0], X_out[y_out == 0, 1],
               s=40, facecolors="none", edgecolors="black", linewidth=1.5, label="Class 1")
ax_out.scatter(X_out[y_out == 1, 0], X_out[y_out == 1, 1],
               s=40, facecolors="none", edgecolors="blue", linewidth=1.5, label="Class 2")
ax_out.set_xlabel("x1")
ax_out.set_ylabel("x2")
ax_out.legend()

display(fig_out)
plt.close(fig_out)
```
