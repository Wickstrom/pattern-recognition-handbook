### Digit classification with SVM

This problem is focused on multi-class classification of handwritten digits
based on 64 features. We load the digits dataset (8x8 images of the digits
0-9), keep a couple of features to start with, and then train an SVM on all
64 features. When you feel confident in your implementation, does performance
change if you use an RBF kernel?

```
from sklearn.datasets import load_digits
```

```
# Digits starter: each sample is an 8x8 image flattened into 64 features.
# Targets are 0-9; below we keep the first two pixels as features so the
# data can be plotted, but for a real SVM you would use all 64 features.
digits_data_ex = load_digits()

X_digits_ex = digits_data_ex.data[:, :2]
y_digits_ex = digits_data_ex.target

fig_digits_ex, ax_digits_ex = plt.subplots(figsize=(6, 6))
sc = ax_digits_ex.scatter(
    X_digits_ex[:, 0], X_digits_ex[:, 1], c=y_digits_ex,
    cmap="tab10", s=40, edgecolor="none",
)
ax_digits_ex.set_xlabel("pixel 1")
ax_digits_ex.set_ylabel("pixel 2")
ax_digits_ex.set_title("Digits shown on just their first two features")
fig_digits_ex.colorbar(sc, ax=ax_digits_ex, label="digit")

display(fig_digits_ex)
plt.close(fig_digits_ex)
```

### Spam classification

The [spambase dataset](https://archive.ics.uci.edu/dataset/94/spambase) is a
real-world binary classification problem: decide whether an email is spam.
Each sample is one email described by 57 features (mostly word and character
frequencies), and the target is 1 (spam) or 0 (not spam). This is a higher
-dimensional problem than the digits above — train an SVM on it and evaluate
how well a linear boundary separates spam from legitimate email.

```
from ucimlrepo import fetch_ucirepo
```

```
# Spambase starter: 57 features per email, binary target 0/1. We load the
# dataset and show the shape and class balance. The 57 features are wide
# (different frequencies/scales), so consider standardising before fitting.
spambase = fetch_ucirepo(id=94)

X_spam = spambase.data.features.to_numpy()
y_spam = spambase.data.targets.to_numpy().ravel()

print("Feature matrix:", X_spam.shape)
print("Number of emails per class (0 = ham, 1 = spam):")
print(np.bincount(y_spam))
```
