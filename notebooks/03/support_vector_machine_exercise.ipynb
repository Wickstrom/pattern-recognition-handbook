### Wine classification with SVM

This problem is focused on multi-class classification of wine types based on 13 features. Below, we have simplified the problem by extracting 2 features. Take this as a starting point, but when you feel confident in your implementation, train your classifier on all 13 features. Does performance change?

```
from sklearn.datasets import load_wine
```

```
# Wine exercise starter: same 2-feature projection as the lecture
# example. Students extend this by fitting an SVM on all 13
# features and comparing multi-class performance.
wine_data_ex = load_wine()

X_wine_ex = wine_data_ex.data[:, :2]
feature_1_name_wine_ex = "alcohol"
feature_2_name_wine_ex = "malic acid"
y_wine_ex = wine_data_ex.target
y_names_wine_ex = np.unique(y_wine_ex)
colors_wine_ex = ["black", "blue", "red"]

fig_wine_ex, ax_wine_ex = plt.subplots(figsize=(5, 5))
for class_value_wine_ex, color_wine_ex in zip(y_names_wine_ex, colors_wine_ex):
    ax_wine_ex.scatter(
        X_wine_ex[y_wine_ex == class_value_wine_ex, 0],
        X_wine_ex[y_wine_ex == class_value_wine_ex, 1],
        s=120, facecolors="none",
        edgecolors=color_wine_ex, linewidth=3.0,
        label=f"Class {class_value_wine_ex+1}",
    )
ax_wine_ex.set_xlabel(feature_1_name_wine_ex)
ax_wine_ex.set_ylabel(feature_2_name_wine_ex)
ax_wine_ex.legend()

display(fig_wine_ex)
plt.close(fig_wine_ex)
```

### The kernel trick: linearly inseparable data

Two circles (one class wrapping around the other) are not linearly separable,
so a linear kernel SVM cannot find a good boundary. Fit one SVM with
`kernel="linear"` and another with `kernel="rbf"` on this data and compare the
decision boundaries and support vectors. The RBF kernel maps the data into a
space where the classes are separable — this is the key idea behind the kernel
trick.


```
from sklearn.datasets import make_circles
from sklearn import svm

X_circles_ex, y_circles_ex = make_circles(
    n_samples=300, factor=0.5, noise=0.1, random_state=0
)

fig_circles_ex, ax_circles_ex = plt.subplots(figsize=(6, 6))
ax_circles_ex.scatter(
    X_circles_ex[y_circles_ex == 0, 0], X_circles_ex[y_circles_ex == 0, 1],
    s=40, facecolors="none", edgecolors="black", linewidth=1.5, label="Class 1",
)
ax_circles_ex.scatter(
    X_circles_ex[y_circles_ex == 1, 0], X_circles_ex[y_circles_ex == 1, 1],
    s=40, facecolors="none", edgecolors="blue", linewidth=1.5, label="Class 2",
)
ax_circles_ex.set_xlabel("x1")
ax_circles_ex.set_ylabel("x2")
ax_circles_ex.legend()

display(fig_circles_ex)
plt.close(fig_circles_ex)
```

### The soft margin: effect of the slack parameter C

The data below is *not* linearly separable — the two overlapping Gaussians
guarantee some training error. Fit an SVM with `kernel="linear"` and compare
`C = 0.01`, `C = 1` and `C = 100`. A small `C` allows many margin violations
(wide, smooth boundary); a large `C` lets almost no training error through
(tight boundary that can overfit). Watch how the number of support vectors and
the choice of *which* points become support vectors changes with `C`.


```
from sklearn import svm

n_slack = 100
mu1_slack = np.array([-0.7, -0.7])
mu2_slack = np.array([0.7, 0.7])
sigma_slack = np.array([[0.6, 0.0], [0.0, 0.6]])

x1_slack = np.random.multivariate_normal(mu1_slack, sigma_slack, n_slack)
x2_slack = np.random.multivariate_normal(mu2_slack, sigma_slack, n_slack)
X_slack = np.vstack([x1_slack, x2_slack])
y_slack = np.concatenate([np.zeros(n_slack), np.ones(n_slack)])

fig_slack, ax_slack = plt.subplots(figsize=(6, 6))
ax_slack.scatter(
    X_slack[y_slack == 0, 0], X_slack[y_slack == 0, 1],
    s=40, facecolors="none", edgecolors="black", linewidth=1.5, label="Class 1",
)
ax_slack.scatter(
    X_slack[y_slack == 1, 0], X_slack[y_slack == 1, 1],
    s=40, facecolors="none", edgecolors="blue", linewidth=1.5, label="Class 2",
)
ax_slack.set_xlabel("x1")
ax_slack.set_ylabel("x2")
ax_slack.legend()

display(fig_slack)
plt.close(fig_slack)
```

### Multiclass SVM with three classes

scikit-learn's SVM handles three or more classes automatically with a
one-vs-one scheme. This synthetic dataset places three Gaussians in a triangle
so the classes meet at a single ambiguity point. Fit an SVM on all three
classes, plot the decision regions, and count how the boundary splits the
three-way overlap. Try both a linear and an RBF kernel.


```
from sklearn import svm

n_tri = 100
sigma_tri = np.array([[0.15, 0.0], [0.0, 0.15]])
mu_tri = [
    np.array([0.0, 1.0]),
    np.array([0.0, -0.6]),
    np.array([1.1, 0.2]),
]

X_tri = np.vstack([
    np.random.multivariate_normal(mu, sigma_tri, n_tri) for mu in mu_tri
])
y_tri = np.concatenate([
    np.full(n_tri, k) for k in range(3)
])
colors_tri = ["black", "blue", "red"]

fig_tri, ax_tri = plt.subplots(figsize=(6, 6))
for k in range(3):
    ax_tri.scatter(
        X_tri[y_tri == k, 0], X_tri[y_tri == k, 1],
        s=40, facecolors="none", edgecolors=colors_tri[k],
        linewidth=1.5, label=f"Class {k+1}",
    )
ax_tri.set_xlabel("x1")
ax_tri.set_ylabel("x2")
ax_tri.legend()

display(fig_tri)
plt.close(fig_tri)
```
